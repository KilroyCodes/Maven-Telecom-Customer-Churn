## Set-up

### Import and read data

```python
import pandas as pd
import numpy as np

zipcode_data = pd.read_csv('telecom_zipcode_population.csv')
customer_data = pd.read_csv('telecom_customer_churn.csv')

customer_data.head()
```

**Check for non-null values**

```python
customer_data.info()
```

## Data Preparation

### Prepare X and Y

We'll separate Joined customers then remove from our dataset

```python
joined_customers = customer_data.loc[customer_data['Customer Status']=='Joined']

customer_data = customer_data.loc[customer_data['Customer Status'].isin(['Churned','Stayed'])]
```

**Prepare Y**

```python
y_customer_data = customer_data['Customer Status'].map(
    {'Churned': 1, 'Stayed':0 })

y_customer_data.head()
```

**Prepare X**

- Remove predictor column: Customer Status
- Remove non-predictor columns: Customer ID, Churn Category, Churn Reason

```python
x_customer_data = customer_data.drop(['Customer Status','Customer ID', 'Churn Category', 'Churn Reason'], axis=1)

x_customer_data.head()
```

### Define Numerical and Categorical Columns

```python
numerical_cols = [cname for cname in x_customer_data.columns
                  if x_customer_data[cname].dtypes in ['int64', 'float64']]
categorical_cols = [cname for cname in x_customer_data.columns
                    if x_customer_data[cname].dtypes == 'object']
```

Since there are 19 categorical values, we'll see about decreasing the amount we use for our model to avoid noise

```python
x_customer_data[categorical_cols].nunique()
```

We'll remove City as there are over 1000 values that would have to be encoded, and we already have the Zip Code column to provide location (prevent redundancy)

```python
x_customer_data.drop('City', axis=1, inplace=True)
categorical_cols.remove('City')
```

### Define Data Preprocessors

```python
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import LabelEncoder
```

```python
categorical_preprocessor = Pipeline(steps=[
                                    ('Imputer', SimpleImputer(strategy='most_frequent')),
                                    ('OHEncoder', OneHotEncoder(handle_unknown='ignore'))])

numerical_preprocessor = SimpleImputer(strategy='median')

preprocessing = ColumnTransformer(transformers=
                                  [('cat', categorical_preprocessor, categorical_cols),
                                   ('num', numerical_preprocessor, numerical_cols)])
```

### Train Test Split

```python
from sklearn.model_selection import train_test_split

x_train, x_valid, y_train, y_valid = train_test_split(
    x_customer_data, y_customer_data, test_size=0.2, random_state=0)
```

## XGBoost

```python
x_train = preprocessing.fit_transform(x_train)
x_valid = preprocessing.transform(x_valid)
```

```python
from xgboost import XGBClassifier

model = XGBClassifier(n_estimators=100,
                      learning_rate=0.05,
                      max_depth=10,
                      early_stopping_rounds=10,
                      eval_metric='logloss',
                      random_state=0)

model.fit(x_train, y_train,
                   eval_set=[(x_valid, y_valid)],
                   verbose=False)

model_score = model.score(x_valid, y_valid)

print(f'Model score: {model_score:.4f}')
```

**To pull churn risk scores:**

```python
model.predict_proba(x_valid)[:,1]
```

**To understand which variables contribute most to churn:**

```python
from matplotlib import pyplot as plt
from xgboost import plot_importance

feature_names = preprocessing.get_feature_names_out()
importances = model.feature_importances_

fig, ax = plt.subplots()
idx = np.argsort(importances)[-10:] 
# [-10:] pulls the last 10 values from an ascending list
bars = plt.barh(np.array(feature_names)[idx], importances[idx])
ax.bar_label(bars)
plt.tight_layout()

plt.show()
```
